```python
with open('./conf/sources.txt', 'r') as sources_conf:
    sources = sources_conf.read().split('\n')
    sources = [source for source in sources if source != '' and not source.startswith('#')]

sources
```

```python
from utils import news_parser

news_content = ""
for source in sources:
  html_content = news_parser.get(source)
  html_body = news_parser.extract_body(html_content)
  if html_body is None:
    print(f"Failed to fetch URL: {source}")
    continue
  page_content = news_parser.exclude_unwanted(html_body)
  page_content = news_parser.fix_links(source, page_content)
  if page_content is None:
    print(f"Failed to fix links: {source}")
    continue
  news_content += news_parser.to_markdown(page_content)

news_content
```
